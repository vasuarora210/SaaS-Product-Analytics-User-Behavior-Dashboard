```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
```

```python
np.random.seed(52)
num_users = 2000
```

```python
#User's Table
users = pd.DataFrame({
    "user_id": range(1, num_users + 1),
    "signup_date": pd.to_datetime('2026-01-01') + 
                   pd.to_timedelta(np.random.randint(0, 365, num_users), unit='D'),
    "country": np.random.choice(['India', 'USA', 'UK', 'Germany'], num_users),
    "acquisition_channel": np.random.choice(['Organic', 'Paid Ads', 'Referral'], num_users)
})
```

```python
#Subsciption Table
plan_types = ['Free', 'Basic', 'Pro']
plan_prices = {'Free': 0, 'Basic': 2000, 'Pro': 5000}
```

```python
subscriptions = users[['user_id', 'signup_date']].copy()
subscriptions['plan_type'] = np.random.choice(plan_types, num_users, p=[0.5, 0.3, 0.2])
subscriptions['start_date'] = subscriptions['signup_date']
subscriptions['monthly_price'] = subscriptions['plan_type'].map(plan_prices)
```

```python
subscriptions['end_date'] = subscriptions['start_date'] + \
    pd.to_timedelta(np.random.randint(30, 180, num_users), unit='D')

churn_mask = np.random.rand(num_users) < 0.3
subscriptions.loc[~churn_mask, 'end_date'] = pd.NaT

subscriptions = subscriptions.drop(columns=['signup_date'])
```

```python
#Event table
events_list = []

for _, row in users.iterrows():
    user = row['user_id']
    signup_date = row['signup_date']
    
    num_events = np.random.randint(5, 30)
    
    for _ in range(num_events):
        event_date = signup_date + pd.to_timedelta(
            np.random.randint(0, 180),  # events within 6 months after signup
            unit='D'
        )
        
        events_list.append({
            "user_id": user,
            "event_date": event_date,
            "event_type": np.random.choice(['login', 'feature_use', 'upgrade', 'cancel']),
            "session_duration": round(np.random.uniform(1, 30), 2)
        })

events = pd.DataFrame(events_list)
```
